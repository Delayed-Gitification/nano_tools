```python
import mappy
import edlib
import random

def generate_test_data():
    """Generates mock Nanopore reads with simulated errors."""
    flank_5 = "GCCACCATGG"
    flank_3 = "TAAGCGGCCG"
    barcode = "ACGTACGTACGT" # 12bp barcode
    
    # Simulate a noisy insert (Barcode surrounded by flanks, with some errors)
    # We'll add a deletion to flank 5 and a substitution to flank 3 to mimic Nanopore
    noisy_flank_5 = "GCCACATGG" # Deleted one 'C'
    noisy_flank_3 = "TAAGCGGTCG" # Substituted 'C' for 'T'
    
    insert = noisy_flank_5 + barcode + noisy_flank_3
    
    # Read 1: Forward strand (insert near the 3' end)
    random_seq_5_prime = "".join(random.choices("ACGT", k=800))
    random_seq_3_prime = "".join(random.choices("ACGT", k=30))
    read_1 = random_seq_5_prime + insert + random_seq_3_prime
    
    # Read 2: Reverse strand (insert near the 5' end, reverse complemented)
    read_2 = mappy.revcomp(read_1)
    
    # Read 3: Garbage read (no barcode)
    read_3 = "".join(random.choices("ACGT", k=900))
    
    return [read_1, read_2, read_3], flank_5, flank_3, barcode

def extract_barcode(read_seq, flank_5, flank_3, search_window=200, max_error_rate=0.2):
    """
    Searches the ends of a read for flanking sequences and extracts the barcode.
    Uses mappy.revcomp to check both orientations.
    """
    seq_len = len(read_seq)
    
    # Define error tolerances based on flank lengths
    err_5 = int(len(flank_5) * max_error_rate)
    err_3 = int(len(flank_3) * max_error_rate)

    # We only want to search the first and last N bases to save time.
    # We slice the read into two distinct windows.
    windows = [
        ("5_prime_end", read_seq[:search_window], 0),
        ("3_prime_end", read_seq[-search_window:], max(0, seq_len - search_window))
    ]

    for orientation in ["forward", "reverse"]:
        seq_to_search = read_seq if orientation == "forward" else mappy.revcomp(read_seq)
        
        # Re-slice windows if we flipped the read
        if orientation == "reverse":
            windows = [
                ("5_prime_end", seq_to_search[:search_window], 0),
                ("3_prime_end", seq_to_search[-search_window:], max(0, seq_len - search_window))
            ]

        for window_name, window_seq, offset in windows:
            # mode="HW" is infix (ignores leading/trailing unaligned bases)
            # task="locations" returns the coordinates of the match
            res_5 = edlib.align(flank_5, window_seq, mode="HW", task="locations", k=err_5)
            res_3 = edlib.align(flank_3, window_seq, mode="HW", task="locations", k=err_3)

            # If both flanks are found within the error tolerance
            if res_5['editDistance'] != -1 and res_3['editDistance'] != -1:
                # edlib returns coordinates as inclusive tuples: (start, end)
                # We take the end of the 5' flank and the start of the 3' flank
                end_5 = res_5['locations'][0][1]
                start_3 = res_3['locations'][-1][0] 

                # Ensure they are in the correct order and not overlapping illegally
                if start_3 > end_5:
                    extracted = window_seq[end_5 + 1 : start_3]
                    return {
                        "barcode": extracted,
                        "orientation": orientation,
                        "location": window_name
                    }
                    
    return None


reads, f5, f3, true_bc = generate_test_data()

print(f"Target Flank 5: {f5}")
print(f"Target Flank 3: {f3}")
print(f"True Barcode:   {true_bc}\n")

for i, read in enumerate(reads):
    print(f"--- Processing Read {i+1} (Length: {len(read)}bp) ---")
    result = extract_barcode(read, f5, f3, search_window=250)
    
    if result:
        print(f"Found Barcode: {result['barcode']}")
        print(f"Orientation:   {result['orientation']}")
        print(f"Found in:      {result['location']}")
    else:
        print("No barcode found.")
    print("")
```

```python
import mappy

def build_aligner_dict(plasmid_references):
    """
    Builds a dictionary of mappy Aligners, one for each unique plasmid.
    
    Args:
        plasmid_references (dict): Format {"barcode": "path/to/plasmid.fasta"}
                                   Alternatively, mappy accepts raw strings via the 'seq' 
                                   argument instead of a file path.
    Returns:
        dict: {"barcode": mappy.Aligner}
    """
    aligners = {}
    
    for barcode, ref_path in plasmid_references.items():
        # Initialise aligner with parameters tuned for short, spliced reads
        aligner = mappy.Aligner(
            ref_path, 
            preset="splice", 
            k=14,               # Increased sensitivity for short sequences
            w=5,                # Denser minimiser sampling
            min_chain_score=25  # Retain shorter alignments
        )
        
        if not aligner:
            raise RuntimeError(f"Failed to initialise aligner for {ref_path}")
            
        aligners[barcode] = aligner
        
    return aligners

def process_read(read_seq, extracted_barcode, aligners_dict):
    """
    Maps a read using the specific aligner dictated by its barcode.
    """
    # 1. Validate barcode against known dictionary
    if extracted_barcode not in aligners_dict:
        # Discard or log reads with unrecognised/mutated barcodes
        return None 
        
    # 2. Retrieve the dedicated aligner
    aligner = aligners_dict[extracted_barcode]
    
    # 3. Perform the alignment
    # aligner.map() returns an iterator of mappy Alignment objects
    alignments = list(aligner.map(read_seq))
    
    if not alignments:
        return None
        
    # In many cases, you only want the primary alignment
    primary_alignment = None
    for aln in alignments:
        if aln.is_primary:
            primary_alignment = aln
            break
            
    return primary_alignment

# --- Example Usage ---

# Mock setup
# In reality, point these to your individual plasmid FASTA files
mock_library = {
    "ACGTACGTACGT": "plasmid_A.fasta",
    "TGCATGCATGCA": "plasmid_B.fasta"
}

# Example of how you would integrate this with the Step 1 logic
print("Building aligner dictionary...")
# my_aligners = build_aligner_dict(mock_library) 
print("Ready to process reads.")

# mock_read = "ACGT..."
# mock_barcode = "ACGTACGTACGT"
# result = process_read(mock_read, mock_barcode, my_aligners)
```

```python
import mappy
import edlib
import pysam
import os
import tempfile
import logging
from dataclasses import dataclass
from enum import Enum
from typing import Optional

logging.basicConfig(level=logging.INFO, format="%(asctime)s [%(levelname)s] %(message)s")
logger = logging.getLogger(__name__)

# --- 1. Configuration & Setup ---

FLANK_5_FWD = "GCCACCATGG"
FLANK_3_FWD = "TAAGCGGCCG"

# On the reverse strand, the 3' flank appears first (5' of the read).
FLANK_5_REV = mappy.revcomp(FLANK_3_FWD)
FLANK_3_REV = mappy.revcomp(FLANK_5_FWD)


class MappingResult(Enum):
    SUCCESS = "success"
    NO_FLANKS_FOUND = "no_flanks_found"
    BARCODE_UNRECOGNISED = "barcode_unrecognised"
    BARCODE_AMBIGUOUS = "barcode_ambiguous"
    MAPPING_FAILED = "mapping_failed"


@dataclass
class PipelineStats:
    processed: int = 0
    no_flanks_found: int = 0
    barcode_unrecognised: int = 0
    barcode_ambiguous: int = 0
    mapping_failed: int = 0
    mapped: int = 0

    def log_summary(self):
        logger.info(
            f"Pipeline complete: {self.processed} reads processed | "
            f"{self.mapped} mapped | "
            f"{self.no_flanks_found} flanks missing | "
            f"{self.barcode_unrecognised} barcode unrecognised | "
            f"{self.barcode_ambiguous} barcode ambiguous | "
            f"{self.mapping_failed} mapping failed"
        )


def build_aligner_dict(plasmid_references: dict[str, str]) -> dict[str, mappy.Aligner]:
    """Initialises splice-aware mappy aligners for each barcode."""
    aligners = {}
    for barcode, ref_path in plasmid_references.items():
        aligner = mappy.Aligner(ref_path, preset="splice", k=14, w=5, min_chain_score=25)
        if not aligner:
            raise RuntimeError(f"Failed to load reference: {ref_path}")
        
        if len(aligner.seq_names) > 1:
            logger.warning(
                f"Reference for barcode '{barcode}' has {len(aligner.seq_names)} sequences; "
                f"only '{aligner.seq_names[0]}' will appear in the BAM header."
            )
        aligners[barcode] = aligner
    return aligners


# --- 2. Barcode Extraction & Matching ---

def extract_barcode_sequence(
    read_seq: str,
    search_window: int = 250,
    max_error_rate: float = 0.2,
    min_barcode_len: int = 4,
) -> Optional[str]:
    """
    Searches the terminal ends of a read for flanking sequences and returns
    the raw extracted sequence between them (forward orientation).
    """
    seq_len = len(read_seq)
    search_window = min(search_window, seq_len)

    # Deduplicate windows if the read is shorter than the search window
    windows = [read_seq[:search_window]]
    if seq_len > search_window:
        windows.append(read_seq[-search_window:])

    err_5 = int(len(FLANK_5_FWD) * max_error_rate)
    err_3 = int(len(FLANK_3_FWD) * max_error_rate)

    orientations = [
        ("forward", FLANK_5_FWD, FLANK_3_FWD),
        ("reverse", FLANK_5_REV, FLANK_3_REV),
    ]

    for window_seq in windows:
        for orientation, f5_seq, f3_seq in orientations:
            res_5 = edlib.align(f5_seq, window_seq, mode="HW", task="locations", k=err_5)
            res_3 = edlib.align(f3_seq, window_seq, mode="HW", task="locations", k=err_3)

            if res_5["editDistance"] == -1 or res_3["editDistance"] == -1:
                continue

            end_5   = res_5["locations"][0][1]
            start_3 = res_3["locations"][-1][0]

            if start_3 <= end_5:
                continue

            extracted = window_seq[end_5 + 1 : start_3]

            if len(extracted) < min_barcode_len:
                continue

            if orientation == "reverse":
                extracted = mappy.revcomp(extracted)

            return extracted

    return None


def identify_library_barcode(
    extracted_seq: str,
    known_barcodes: list[str],
    max_edits: int = 2,
) -> tuple[Optional[str], MappingResult]:
    """
    Matches the extracted sequence against the known library barcodes using
    global (NW) alignment to tolerate Nanopore indel errors.
    """
    best_dist = max_edits
    best_matches: list[str] = []

    for bc in known_barcodes:
        # Pass current best_dist as k so edlib can short-circuit worse candidates
        res = edlib.align(extracted_seq, bc, mode="NW", k=best_dist)
        dist = res["editDistance"]

        if dist == -1:
            continue

        if dist < best_dist:
            best_dist = dist
            best_matches = [bc]
        elif dist == best_dist:
            best_matches.append(bc)

    if not best_matches:
        return None, MappingResult.BARCODE_UNRECOGNISED

    if len(best_matches) > 1:
        logger.debug(
            f"Ambiguous barcode: '{extracted_seq}' matches {best_matches} "
            f"at edit distance {best_dist}"
        )
        return None, MappingResult.BARCODE_AMBIGUOUS

    return best_matches[0], MappingResult.SUCCESS


# --- 3. BAM Handling ---

def create_bam_header(
    aligners_dict: dict[str, mappy.Aligner]
) -> tuple[dict, dict[str, int]]:
    """Builds the unified SAM header from the dictionary of aligners."""
    header = {"HD": {"VN": "1.0", "SO": "unsorted"}, "SQ": []}
    ref_name_to_id: dict[str, int] = {}

    for i, (barcode, aligner) in enumerate(sorted(aligners_dict.items())):
        ref_name = aligner.seq_names[0]
        ref_len  = aligner.seq_lens[0]
        header["SQ"].append({"SN": ref_name, "LN": ref_len})
        ref_name_to_id[ref_name] = i

    return header, ref_name_to_id


def _build_aligned_segment(
    bam_writer: pysam.AlignmentFile,
    ref_id_map: dict[str, int],
    read_name: str,
    read_seq: str,
    aln: mappy.Alignment,
    cached_qual_array,
) -> pysam.AlignedSegment:
    """Constructs a pysam AlignedSegment, handling chimeras correctly."""
    a = pysam.AlignedSegment(bam_writer.header)
    a.query_name      = read_name
    a.reference_id    = ref_id_map[aln.ctg]
    a.reference_start = aln.r_st
    a.mapping_quality = aln.mapq
    a.cigar = [(op, length) for length, op in aln.cigar]

    a.is_supplementary = not aln.is_primary
    a.is_reverse       = (aln.strand == -1)

    if a.is_supplementary:
        a.query_sequence  = None
        a.query_qualities = None
    else:
        if a.is_reverse:
            a.query_sequence = mappy.revcomp(read_seq)
            if cached_qual_array is not None:
                a.query_qualities = cached_qual_array[::-1]
        else:
            a.query_sequence = read_seq
            if cached_qual_array is not None:
                a.query_qualities = cached_qual_array

    return a


def process_and_write_read(
    bam_writer: pysam.AlignmentFile,
    ref_id_map: dict[str, int],
    read_name: str,
    read_seq: str,
    read_qual: Optional[str],
    barcode: str,
    aligners_dict: dict[str, mappy.Aligner],
) -> MappingResult:
    """Maps the read to its designated plasmid and writes all alignment hits."""
    aligner = aligners_dict.get(barcode)
    if not aligner:
        return MappingResult.BARCODE_UNRECOGNISED

    alignments = list(aligner.map(read_seq))
    if not alignments:
        return MappingResult.MAPPING_FAILED

    cached_qual_array = pysam.qualitystring_to_array(read_qual) if read_qual else None

    for aln in alignments:
        segment = _build_aligned_segment(
            bam_writer, ref_id_map, read_name, read_seq, aln, cached_qual_array
        )
        bam_writer.write(segment)

    return MappingResult.SUCCESS


# --- 4. Main Execution ---

def run_pipeline(
    fastq_path: str,
    plasmid_library: dict[str, str],
    output_bam_path: str,
    barcode_max_edits: int = 2,
    log_interval: int = 10_000,
) -> PipelineStats:
    
    logger.info("Initialising aligners...")
    aligners = build_aligner_dict(plasmid_library)
    header, ref_map = create_bam_header(aligners)
    known_barcodes = sorted(plasmid_library.keys())

    stats = PipelineStats()
    temp_bam_fd, temp_bam_path = tempfile.mkstemp(suffix=".bam")
    os.close(temp_bam_fd)

    try:
        logger.info("Processing reads...")
        with pysam.AlignmentFile(temp_bam_path, "wb", header=header) as unsorted_bam:
            for name, seq, qual in mappy.fastx_read(fastq_path):

                stats.processed += 1
                if stats.processed % log_interval == 0:
                    logger.info(f"Processed {stats.processed} reads...")

                raw_extracted = extract_barcode_sequence(seq)
                if not raw_extracted:
                    stats.no_flanks_found += 1
                    continue

                matched_barcode, match_result = identify_library_barcode(
                    raw_extracted, known_barcodes, max_edits=barcode_max_edits
                )
                
                if matched_barcode is None:
                    match match_result:
                        case MappingResult.BARCODE_UNRECOGNISED:
                            stats.barcode_unrecognised += 1
                        case MappingResult.BARCODE_AMBIGUOUS:
                            stats.barcode_ambiguous += 1
                    continue

                result = process_and_write_read(
                    unsorted_bam, ref_map, name, seq, qual, matched_barcode, aligners
                )
                
                match result:
                    case MappingResult.SUCCESS:
                        stats.mapped += 1
                    case MappingResult.MAPPING_FAILED:
                        stats.mapping_failed += 1
                    case MappingResult.BARCODE_UNRECOGNISED:
                        stats.barcode_unrecognised += 1

        logger.info("Sorting and indexing final BAM file...")
        pysam.sort("-o", output_bam_path, temp_bam_path)
        pysam.index(output_bam_path)

    finally:
        if os.path.exists(temp_bam_path):
            os.remove(temp_bam_path)

    stats.log_summary()
    return stats



# Example usage:
# mock_library = {
#     "ACGTACGTACGT": "refs/plasmid_A.fasta",
#     "TGCATGCATGCA": "refs/plasmid_B.fasta",
# }
# stats = run_pipeline(
#     "data/nanopore_reads.fastq.gz",
#     mock_library,
#     "results/final_sorted.bam",
#     barcode_max_edits=2,
# )
```
