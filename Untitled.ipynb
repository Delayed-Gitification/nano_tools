```python
import torch
import numpy as np
from transformers import AutoModelForCausalLM, AutoTokenizer
import time

class HyenaDNAEncoder:
    def __init__(self, model_name="LongSafari/hyenadna-tiny-1k-seqlen-hf"):
        print(f"🚀 Loading {model_name}...")
        self.tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
        self.model = AutoModelForCausalLM.from_pretrained(model_name, trust_remote_code=True)
        self.model.eval()
        
        # Critical Fix: Ensure pad token is defined for the manual mask logic
        if self.tokenizer.pad_token is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token
            self.tokenizer.pad_token_id = self.tokenizer.eos_token_id

    def encode_batch(self, sequences, batch_size=32):
        all_vectors = []
        
        for i in range(0, len(sequences), batch_size):
            batch_seqs = sequences[i : i + batch_size]
            
            # 1. Tokenize
            inputs = self.tokenizer(
                batch_seqs, 
                return_tensors="pt", 
                padding=True, 
                truncation=True, 
                max_length=1024 
            )
            
            # 2. Manual Mask Generation (Fixes KeyError)
            if "attention_mask" in inputs:
                attention_mask = inputs["attention_mask"]
            else:
                # Create mask manually: 1 for real tokens, 0 for padding
                attention_mask = (inputs["input_ids"] != self.tokenizer.pad_token_id).long()
            
            # 3. Inference
            with torch.no_grad():
                outputs = self.model(**inputs, output_hidden_states=True)
                
                # Layer -1 is the last layer
                last_hidden_state = outputs.hidden_states[-1]
                
                # 4. Mean Pooling with the safe mask
                # Expand mask to match hidden state dimensions: (Batch, Len) -> (Batch, Len, 1)
                mask_expanded = attention_mask.unsqueeze(-1).float()
                
                sum_embeddings = torch.sum(last_hidden_state * mask_expanded, dim=1)
                sum_mask = torch.clamp(mask_expanded.sum(dim=1), min=1e-9)
                
                mean_pooled = sum_embeddings / sum_mask
                all_vectors.append(mean_pooled.numpy())

        return np.vstack(all_vectors)

# --- TEST BENCH ---
print("Generating dummy data...")
# Note: HyenaDNA-Tiny max length is 1024. 
sequences = ["ACGT" * 50 for _ in range(100)] # 200bp sequences

print("Initializing...")
encoder = HyenaDNAEncoder()

print("Encoding...")
start = time.time()
vectors = encoder.encode_batch(sequences)
end = time.time()

print(f"✅ Success! Processed {len(sequences)} sequences in {end - start:.2f} seconds.")
print(f"Output Shape: {vectors.shape}")
```

```python
import torch
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import cosine_similarity
from transformers import AutoModelForCausalLM, AutoTokenizer
from sklearn.decomposition import PCA

# --- 1. ENCODER CLASS (Standard) ---
class HyenaDNAEncoder:
    def __init__(self, model_name="LongSafari/hyenadna-tiny-1k-seqlen-hf"):
        print(f"🚀 Loading {model_name}...")
        self.tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
        self.model = AutoModelForCausalLM.from_pretrained(model_name, trust_remote_code=True)
        self.model.eval()
        if self.tokenizer.pad_token is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token
            self.tokenizer.pad_token_id = self.tokenizer.eos_token_id

    def encode_batch(self, sequences, batch_size=32):
        all_vectors = []
        for i in range(0, len(sequences), batch_size):
            batch_seqs = sequences[i : i + batch_size]
            inputs = self.tokenizer(batch_seqs, return_tensors="pt", padding=True, truncation=True, max_length=1024)
            
            if "attention_mask" in inputs:
                attention_mask = inputs["attention_mask"]
            else:
                attention_mask = (inputs["input_ids"] != self.tokenizer.pad_token_id).long()
            
            with torch.no_grad():
                outputs = self.model(**inputs, output_hidden_states=True)
                last_hidden_state = outputs.hidden_states[-1]
                mask_expanded = attention_mask.unsqueeze(-1).float()
                sum_embeddings = torch.sum(last_hidden_state * mask_expanded, dim=1)
                sum_mask = torch.clamp(mask_expanded.sum(dim=1), min=1e-9)
                mean_pooled = sum_embeddings / sum_mask
                all_vectors.append(mean_pooled.numpy())
        return np.vstack(all_vectors)

# --- 2. CLEANING FUNCTION ---
def clean_embeddings(embeddings, method='center_only', n_components=None):
    n_samples = embeddings.shape[0]
    if n_components is not None and n_components > n_samples:
        n_components = n_samples

    if method == 'center_only':
        mu = np.mean(embeddings, axis=0)
        return embeddings - mu
    elif method == 'pca':
        pca = PCA(n_components=n_components, whiten=False)
        return pca.fit_transform(embeddings)
    elif method == 'whitening':
        pca = PCA(n_components=n_components, whiten=True)
        return pca.fit_transform(embeddings)
    return embeddings

# --- 3. NEW DATA GENERATION: 3 RANDOM FAMILIES ---
def mutate(seq, rate=0.2):
    chars = list(seq)
    for i in range(len(chars)):
        if np.random.rand() < rate:
            chars[i] = np.random.choice(list("ACGT"))
    return "".join(chars)

print("Generating 3 distinct random families...")

# Create 3 unique ancestors
ancestor_1 = "".join(np.random.choice(list("ACGT"), 200))
ancestor_2 = "".join(np.random.choice(list("ACGT"), 200))
ancestor_3 = "".join(np.random.choice(list("ACGT"), 200))

# Generate 20 variations for each ancestor (Rate 0.2 = 20% difference)
group_1 = [ancestor_1] + [mutate(ancestor_1, rate=0.1) for _ in range(19)]
group_2 = [ancestor_2] + [mutate(ancestor_2, rate=0.1) for _ in range(19)]
group_3 = [ancestor_3] + [mutate(ancestor_3, rate=0.1) for _ in range(19)]

all_seqs = group_1 + group_2 + group_3
labels = ["Family A"]*20 + ["Family B"]*20 + ["Family C"]*20

# --- 4. EXECUTION ---
encoder = HyenaDNAEncoder()
print("Encoding...")
raw_embeddings = encoder.encode_batch(all_seqs)

# Apply methods
centered = clean_embeddings(raw_embeddings, method='center_only')
whitened = clean_embeddings(raw_embeddings, method='whitening', n_components=32)

# --- 5. VISUALIZATION ---
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# 1. Raw
sim_raw = cosine_similarity(raw_embeddings)
sns.heatmap(sim_raw, ax=axes[0], cmap="viridis", vmin=0.9, vmax=1.0)
axes[0].set_title("Raw Embeddings\n(Everything looks same)")

# 2. Centered (Best)
sim_centered = cosine_similarity(centered)
sns.heatmap(sim_centered, ax=axes[1], cmap="coolwarm", center=0)
axes[1].set_title("Centered (Best)\n(Distinct Families)")

# 3. Whitened (Too Strong)
sim_whitened = cosine_similarity(whitened)
sns.heatmap(sim_whitened, ax=axes[2], cmap="coolwarm", center=0)
axes[2].set_title("Whitened\n(Clusters Destroyed)")

plt.tight_layout()
plt.show()

# --- 6. PRINT METRICS ---
print("-" * 30)
print(f"Family A vs Family A (Intra-cluster): {sim_centered[0, 1]:.4f}")
print(f"Family A vs Family B (Inter-cluster): {sim_centered[0, 25]:.4f}")
```

```python
import numpy as np
import itertools
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.decomposition import PCA

# --- 1. THE SPATIAL K-MER ENCODER ---
class SpatiallyPooledKmerEncoder:
    def __init__(self, k=6, n_bins=4):
        self.k = k
        self.n_bins = n_bins
        # Generate all possible k-mers (e.g., AAA, AAC, ... TTT)
        self.vocab = ["".join(p) for p in itertools.product("ACGT", repeat=k)]
        self.vocab_map = {kmer: i for i, kmer in enumerate(self.vocab)}
        self.vocab_size = len(self.vocab)
        print(f"🔧 Initialized K-mer Encoder (k={k}, bins={n_bins}, feature_dim={self.vocab_size * n_bins})")

    def encode_batch(self, sequences):
        feature_vectors = []
        
        for seq in sequences:
            seq_len = len(seq)
            # Calculate bin size (roughly)
            bin_size = int(np.ceil(seq_len / self.n_bins))
            
            bin_vectors = []
            
            for b in range(self.n_bins):
                # Slicing the sequence for this bin
                start = b * bin_size
                end = min(start + bin_size, seq_len)
                segment = seq[start:end]
                
                # Count K-mers in this segment
                counts = np.zeros(self.vocab_size)
                
                # Iterate through segment
                if len(segment) >= self.k:
                    for i in range(len(segment) - self.k + 1):
                        kmer = segment[i : i+self.k]
                        if kmer in self.vocab_map:
                            idx = self.vocab_map[kmer]
                            counts[idx] += 1
                
                # Normalize (L1 Norm / Frequency) - Vital for comparison!
                # If we don't normalize, longer bins would dominate
                total = counts.sum()
                if total > 0:
                    counts = counts / total
                    
                bin_vectors.append(counts)
            
            # Concatenate all bin vectors into one long vector
            full_vector = np.concatenate(bin_vectors)
            feature_vectors.append(full_vector)
            
        return np.vstack(feature_vectors)

# --- 2. HELPERS (Reuse from previous) ---
def clean_embeddings(embeddings, method='center_only'):
    # Even K-mer counts benefit from centering to remove "common background"
    if method == 'center_only':
        return embeddings - np.mean(embeddings, axis=0)
    return embeddings

def mutate(seq, rate=0.2):
    chars = list(seq)
    for i in range(len(chars)):
        if np.random.rand() < rate:
            chars[i] = np.random.choice(list("ACGT"))
    return "".join(chars)

# --- 3. DATA GENERATION (3 FAMILIES) ---
print("Generating 3 distinct random families...")
# Ancestors
anc_1 = "".join(np.random.choice(list("ACGT"), 200))
anc_2 = "".join(np.random.choice(list("ACGT"), 200))
anc_3 = "".join(np.random.choice(list("ACGT"), 200))

# Descendants (20% mutation rate)
group_1 = [mutate(anc_1, rate=0.05) for _ in range(19)]
group_2 = [mutate(anc_2, rate=0.05) for _ in range(19)]
group_3 = [mutate(anc_3, rate=0.05) for _ in range(19)]

all_seqs = group_1 + group_2 + group_3

# --- 4. EXECUTION ---
# Using k=3 (64 dims) and 4 bins = 256 total dimensions
encoder = SpatiallyPooledKmerEncoder(k=6, n_bins=10)

print("Encoding...")
raw_vectors = encoder.encode_batch(all_seqs)

# Apply Centering (Crucial for cleaner heatmaps)
centered_vectors = clean_embeddings(raw_vectors, method='center_only')

# --- 5. VISUALIZATION ---
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot 1: Raw Counts (Usually dominated by common k-mers)
sns.heatmap(cosine_similarity(raw_vectors), ax=axes[0], cmap="viridis")
axes[0].set_title("Raw K-mer Frequency\n(Note: Can be dominated by 'AAA')")

# Plot 2: Centered (Removes common background frequencies)
sim_centered = cosine_similarity(centered_vectors)
sns.heatmap(sim_centered, ax=axes[1], cmap="coolwarm", center=0)
axes[1].set_title("Centered K-mer Vectors\n(Clean Clusters!)")

plt.tight_layout()
plt.show()

# --- 6. METRICS ---
print("-" * 30)
print(f"Family A Intra-Cluster: {sim_centered[0, 1]:.4f}")
print(f"Family A vs Family B:   {sim_centered[0, 25]:.4f}")
```
