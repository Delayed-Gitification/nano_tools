```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import kurtosis  # <--- NEW IMPORT
from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, silhouette_samples
from matplotlib.colors import ListedColormap
import random

# --- 1. Helper Functions ---

def encode_msa(msa_strings, alphabet="ACGTN-"):
    """Converts list of MSA strings to an integer matrix."""
    char_to_int = {c: i for i, c in enumerate(alphabet)}
    msa_array = np.array([list(seq) for seq in msa_strings])
    N, L = msa_array.shape
    
    msa_int = np.zeros((N, L), dtype=np.int8)
    for char, idx in char_to_int.items():
        msa_int[msa_array == char] = idx
        
    return msa_int, len(alphabet)

def get_marginals(msa_int, vocab_size):
    """Returns One-Hot encoding (N, L, A) and Marginal Probs P_i (L, A)."""
    one_hot = np.eye(vocab_size)[msa_int] 
    P_i = one_hot.mean(axis=0)            
    return one_hot, P_i

def compute_mi_scores(one_hot, P_i):
    """Calculates the MI Matrix using pre-computed One-Hot and Marginals."""
    N, L, A = one_hot.shape
    flat_view = one_hot.transpose(1, 2, 0).reshape(L * A, N)
    joint_probs_flat = (flat_view @ flat_view.T) / N
    P_ij = joint_probs_flat.reshape(L, A, L, A)
    P_product = P_i[:, :, None, None] * P_i[None, None, :, :]
    mask = P_ij > 0
    mi_matrix = np.zeros_like(P_ij)
    mi_matrix[mask] = P_ij[mask] * np.log(P_ij[mask] / P_product[mask])
    return mi_matrix.sum(axis=(1, 3))

def get_all_metrics(mi_matrix, exclusion_distance=5):
    """Extracts Total MI, Max Row, Max Pair, and Top 1% Ratio."""
    L = mi_matrix.shape[0]
    
    # Create mask (Upper Triangle, ignoring diagonal & exclusion zone)
    mask = np.triu(np.ones((L, L), dtype=bool), k=exclusion_distance + 1)
    valid_scores = mi_matrix[mask]
    
    # 1. Total MI
    total_mi = valid_scores.sum()
    
    # 2. Max Pair
    max_pair = valid_scores.max() if len(valid_scores) > 0 else 0
    
    # 3. Max Row Sum
    masked_matrix = np.zeros_like(mi_matrix)
    masked_matrix[mask] = mi_matrix[mask]
    sym_masked = masked_matrix + masked_matrix.T
    max_row_sum = sym_masked.sum(axis=1).max()
    
    # 4. Top 1% Ratio (Signal-to-Noise Ratio)
    if len(valid_scores) > 0:
        global_mean = valid_scores.mean()
        
        # Avoid divide by zero if matrix is empty/zero
        if global_mean > 1e-9:
            # Determine how many items constitute the top 1% (at least 1)
            n_top = max(1, int(len(valid_scores) * 0.01))
            
            # Sort and grab the top slice
            # (partition is faster than sort for finding top k, but sort is safer for means)
            top_scores = np.sort(valid_scores)[-n_top:]
            top_mean = top_scores.mean()
            
            ratio_val = top_mean / global_mean
        else:
            ratio_val = 0
    else:
        ratio_val = 0
    
    return total_mi, max_row_sum, max_pair, ratio_val

# --- 2. Main Simulation Function (Updated) ---

def run_outlier_test(msa_int, vocab_size, n_sims=1000, exclusion_distance=5):
    """Runs Monte Carlo sim to estimate p-values including Top 1% Ratio."""
    N, L = msa_int.shape
    
    real_one_hot, P_i = get_marginals(msa_int, vocab_size)
    real_mi_matrix = compute_mi_scores(real_one_hot, P_i)
    # UNPACK RATIO
    r_total, r_row, r_pair, r_ratio = get_all_metrics(real_mi_matrix, exclusion_distance)
    
    n_total, n_row, n_pair, n_ratio = [], [], [], []
    shuffled_int = msa_int.copy()
    
    for i in range(n_sims):
        for col in range(L):
            np.random.shuffle(shuffled_int[:, col])
            
        shuffled_one_hot, _ = get_marginals(shuffled_int, vocab_size)
        sim_mi_matrix = compute_mi_scores(shuffled_one_hot, P_i)
        
        # UNPACK RATIO
        s_total, s_row, s_pair, s_ratio = get_all_metrics(sim_mi_matrix, exclusion_distance)
        n_total.append(s_total)
        n_row.append(s_row)
        n_pair.append(s_pair)
        n_ratio.append(s_ratio)

    p_total = (np.sum(np.array(n_total) >= r_total) + 1) / (n_sims + 1)
    p_row   = (np.sum(np.array(n_row)   >= r_row)   + 1) / (n_sims + 1)
    p_pair  = (np.sum(np.array(n_pair)  >= r_pair)  + 1) / (n_sims + 1)
    p_ratio = (np.sum(np.array(n_ratio) >= r_ratio) + 1) / (n_sims + 1)

    return {
        "p_values": (p_total, p_row, p_pair, p_ratio),
        "real_mi_matrix": real_mi_matrix
    }

# --- 3. Splitting Logic (Unchanged from previous fix) ---

def get_elbow_columns(mi_matrix, exclusion_distance=5):
    L = mi_matrix.shape[0]
    mask = np.triu(np.ones((L, L), dtype=bool), k=exclusion_distance + 1)
    rows, cols = np.where(mask)
    scores = mi_matrix[rows, cols]
    
    sorted_indices = np.argsort(scores)[::-1]
    cap = min(100, len(sorted_indices))
    sorted_indices = sorted_indices[:cap]
    sorted_scores = scores[sorted_indices]
    
    if len(sorted_scores) < 3:
        sel_rows = rows[sorted_indices]
        sel_cols = cols[sorted_indices]
        return np.unique(np.concatenate([sel_rows, sel_cols]))

    n_points = len(sorted_scores)
    x_norm = np.linspace(0, 1, n_points)
    y_norm = (sorted_scores - sorted_scores.min()) / (sorted_scores.max() - sorted_scores.min() + 1e-9)
    
    line_vec = np.array([1, y_norm[-1] - y_norm[0]])
    line_vec = line_vec / np.linalg.norm(line_vec)
    vec_from_start = np.stack([x_norm, y_norm - y_norm[0]], axis=1)
    distances = np.abs(vec_from_start[:, 0] * line_vec[1] - vec_from_start[:, 1] * line_vec[0])
    
    elbow_idx = np.argmax(distances)
    n_selected = max(1, elbow_idx // 2)  # Conservative selection
    selected_indices = sorted_indices[:n_selected]
    
    sel_rows = rows[selected_indices]
    sel_cols = cols[selected_indices]
    return np.unique(np.concatenate([sel_rows, sel_cols]))

def perform_pca_gmm_split(msa_int, vocab_size, col_indices, confidence_threshold=0.90):
    msa_subset = msa_int[:, col_indices]
    subset_one_hot = np.eye(vocab_size)[msa_subset]
    features = subset_one_hot.reshape(msa_int.shape[0], -1)
    
    n_comps = min(2, features.shape[1])
    pca = PCA(n_components=n_comps)
    pca_coords = pca.fit_transform(features)
    
    gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)
    gmm.fit(pca_coords)
    
    probs = gmm.predict_proba(pca_coords) 
    max_probs = np.max(probs, axis=1)
    labels = gmm.predict(pca_coords)
    
    if len(np.unique(labels)) < 2:
        ambiguous_mask = max_probs < confidence_threshold
    else:
        sil_samples = silhouette_samples(pca_coords, labels)
        ambiguous_mask = (max_probs < confidence_threshold) | (sil_samples < 0.05)

    labels[ambiguous_mask] = -1 
    
    valid_mask = ~ambiguous_mask
    valid_labels = labels[valid_mask]
    valid_coords = pca_coords[valid_mask]
    
    if len(valid_labels) > 0 and len(np.unique(valid_labels)) >= 2:
        sil_score = silhouette_score(valid_coords, valid_labels)
    else:
        sil_score = -1 
    
    return labels, sil_score, pca_coords

def visualize_split(msa_int, labels, col_indices, alphabet="ACGTN-"):
    msa_subset = msa_int[:, col_indices]
    sort_idx = np.argsort(labels)
    sorted_msa = msa_subset[sort_idx]
    sorted_labels = labels[sort_idx]
    
    colors = ['#32CD32', '#4169E1', '#FFA500', '#DC143C', '#808080', '#FFFFFF']
    cmap_dna = ListedColormap(colors[:len(alphabet)])

    fig, (ax_lbl, ax_msa) = plt.subplots(1, 2, figsize=(12, 6), 
                                         gridspec_kw={'width_ratios': [1, 20], 'wspace': 0.05})
    
    label_cmap = ListedColormap(['#d3d3d3', '#1f77b4', '#ff7f0e']) 
    ax_lbl.imshow(sorted_labels[:, None], aspect='auto', cmap=label_cmap, interpolation='nearest')
    ax_lbl.set_title("ID", fontsize=10)
    ax_lbl.axis('off')
    
    im = ax_msa.imshow(sorted_msa, aspect='auto', interpolation='nearest', cmap=cmap_dna, vmin=0, vmax=len(alphabet)-1)
    
    ax_msa.set_title(f"Informative Columns Only ({len(col_indices)} cols)", fontsize=12)
    ax_msa.set_xlabel("Original Column Index")
    ax_msa.set_xticks(np.arange(len(col_indices)))
    ax_msa.set_xticklabels(col_indices, rotation=90, fontsize=8)
    ax_msa.set_yticks([]) 
    
    cbar = plt.colorbar(im, ax=ax_msa, ticks=np.arange(len(alphabet)))
    cbar.ax.set_yticklabels(list(alphabet))
    
    plt.tight_layout()
    plt.show()

# --- 4. Recursive Pipeline (Updated with Kurtosis) ---

def recursive_msa_split(msa_strings, p_threshold=0.01, n_sims=1000, alphabet="ACGTN-", exclusion_distance=5):
    if len(msa_strings) < 3:
        return [msa_strings]

    msa_int, vocab_size = encode_msa(msa_strings, alphabet)
    stats = run_outlier_test(msa_int, vocab_size, n_sims=n_sims, exclusion_distance=exclusion_distance)
    
    # UNPACK 4 P-VALUES NOW
    p_total, p_row, p_pair, p_ratio = stats["p_values"]
    print(f"Analyzing {len(msa_strings)} seqs -> P-values: Total={p_total:.3f}, Row={p_row:.3f}, Pair={p_pair:.3f}, Ratio={p_ratio:.3f}")
    
    # DECISION: Check if Ratio AND Max Pair are significant
    if (p_ratio > p_threshold) and (p_pair > p_threshold):
        print(f"-> No significant signal. Returning cluster.")
        return [msa_strings]
    print(f"-> SIGNAL DETECTED. Splitting...")
    top_cols = get_elbow_columns(stats["real_mi_matrix"], exclusion_distance=exclusion_distance)
    print(f"   Elbow selected {len(top_cols)} unique informative columns.")
    
    labels, score, _ = perform_pca_gmm_split(msa_int, vocab_size, top_cols, confidence_threshold=0.90)
    print(f"   Split Silhouette (Valid Reads): {score:.3f}")

    visualize_split(msa_int, labels, top_cols, alphabet)
    
    group_A = [s for i, s in enumerate(msa_strings) if labels[i] == 0]
    group_B = [s for i, s in enumerate(msa_strings) if labels[i] == 1]
    
    n_dropped = len(msa_strings) - (len(group_A) + len(group_B))
    if n_dropped > 0:
        print(f"   (Dropped {n_dropped} ambiguous reads)")
    
    results = []
    results.extend(recursive_msa_split(group_A, p_threshold, n_sims, alphabet, exclusion_distance))
    results.extend(recursive_msa_split(group_B, p_threshold, n_sims, alphabet, exclusion_distance))
    
    return results

# --- Usage ---
final_clusters = recursive_msa_split(msa, p_threshold=0.01, n_sims=100, exclusion_distance=1)

def mutate_sequence(seq, error_rate=0.10):
    if error_rate <= 0: return seq
    seq_list = list(seq)
    new_seq = []
    for base in seq_list:
        if random.random() < error_rate:
            r = random.random()
            if r < 0.5: new_seq.append(random.choice("ACGT")) 
            elif r < 0.75: 
                new_seq.append(base)
                new_seq.append(random.choice("ACGT"))
            else: pass 
        else:
            new_seq.append(base)
    return "".join(new_seq)


plas1 = "ACGGGATCATCGACTTTGCACACTATGGGACTAGCACTAGCACTGACACTGCCGCGCGCGGGGGCGCCACTATTTACTACGCA"

for _ in range(20):
    plas2 = mutate_sequence(plas1, 0.05)
    
    plas1_bc = "ACGAAAAGCGCTTT"
    plas2_bc = "ACGAATAGCGCTTT"
    
    n1 = 10
    n2 = 10
    
    plas1_seqs = [mutate_sequence(plas1, 0.05) for _ in range(n1)]
    plas2_seqs = [mutate_sequence(plas2, 0.05) for _ in range(n2)]
    
    plas1_bcs = [mutate_sequence(plas1_bc, 0.05) for _ in range(n1)]
    plas2_bcs = [mutate_sequence(plas2_bc, 0.05) for _ in range(n2)]
    
    plas_seqs = plas1_seqs + plas2_seqs
    bc_seqs = plas1_bcs + plas2_bcs
    
    from spoa import poa
    _, msa_plasmids = poa(plas_seqs)
    
    bc_gap_penalty = -20
    
    _, msa_bcs = poa(bc_seqs, g=bc_gap_penalty, algorithm=2)
    msa_bcs
    
    msa = [a + "-"*10 + b for a, b in zip(msa_bcs, msa_plasmids)]

    final_clusters = recursive_msa_split(msa, p_threshold=0.05, n_sims=100, exclusion_distance=15)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, silhouette_samples
from matplotlib.colors import ListedColormap
import random

# --- 1. Helper Functions ---

def encode_msa(msa_strings, alphabet="ACGTN-"):
    """Converts list of MSA strings to an integer matrix."""
    char_to_int = {c: i for i, c in enumerate(alphabet)}
    msa_array = np.array([list(seq) for seq in msa_strings])
    N, L = msa_array.shape
    
    msa_int = np.zeros((N, L), dtype=np.int8)
    for char, idx in char_to_int.items():
        msa_int[msa_array == char] = idx
        
    return msa_int, len(alphabet)

def get_marginals(msa_int, vocab_size):
    """Returns One-Hot encoding (N, L, A) and Marginal Probs P_i (L, A)."""
    one_hot = np.eye(vocab_size)[msa_int] 
    P_i = one_hot.mean(axis=0)            
    return one_hot, P_i

def compute_mi_scores(one_hot, P_i):
    """Calculates the MI Matrix using pre-computed One-Hot and Marginals."""
    N, L, A = one_hot.shape
    flat_view = one_hot.transpose(1, 2, 0).reshape(L * A, N)
    joint_probs_flat = (flat_view @ flat_view.T) / N
    P_ij = joint_probs_flat.reshape(L, A, L, A)
    P_product = P_i[:, :, None, None] * P_i[None, None, :, :]
    mask = P_ij > 0
    mi_matrix = np.zeros_like(P_ij)
    mi_matrix[mask] = P_ij[mask] * np.log(P_ij[mask] / P_product[mask])
    return mi_matrix.sum(axis=(1, 3))

def calculate_entropy_from_counts(counts, n_rows):
    """
    Calculates total entropy (sum across all columns) given a counts matrix (L, A).
    Vectorized for speed.
    """
    if n_rows == 0: return 0.0
    
    # Convert counts to probabilities
    probs = counts / n_rows
    
    # Calculate -p * log2(p), handling 0s
    log_probs = np.zeros_like(probs)
    mask = probs > 0
    log_probs[mask] = np.log2(probs[mask])
    
    # Sum over characters (axis 1) then over columns (axis 0)
    return -np.sum(probs * log_probs)

def check_split_entropy_significance(msa_int, vocab_size, labels, n_perms=100):
    """
    1. Calculates Information Gain (IG) of the proposed split.
    2. Shuffles labels 100 times (preserving group sizes) to generate Null IG.
    3. Returns the P-value.
    """
    # Filter out 'ambiguous' (-1) reads for this calculation
    valid_mask = labels != -1
    valid_labels = labels[valid_mask]
    valid_msa = msa_int[valid_mask]
    
    N, L = valid_msa.shape
    # If split is degenerate (0 reads in one group), it's invalid
    if N < 4 or len(np.unique(valid_labels)) < 2: 
        return 1.0 
    
    # Pre-calculate One-Hot encoding
    one_hot = np.eye(vocab_size)[valid_msa]
    
    # --- A. Calculate Real IG ---
    parent_counts = one_hot.sum(axis=0)
    parent_H = calculate_entropy_from_counts(parent_counts, N)
    
    def get_weighted_child_entropy(lbls):
        mask0 = lbls == 0
        n0 = np.sum(mask0)
        counts0 = one_hot[mask0].sum(axis=0)
        H0 = calculate_entropy_from_counts(counts0, n0)
        
        mask1 = lbls == 1
        n1 = np.sum(mask1)
        counts1 = one_hot[mask1].sum(axis=0)
        H1 = calculate_entropy_from_counts(counts1, n1)
        
        return (n0/N)*H0 + (n1/N)*H1

    real_child_H = get_weighted_child_entropy(valid_labels)
    real_gain = parent_H - real_child_H
    
    # --- B. Permutation Test ---
    null_gains = []
    perm_labels = valid_labels.copy()
    
    for _ in range(n_perms):
        np.random.shuffle(perm_labels)
        perm_child_H = get_weighted_child_entropy(perm_labels)
        perm_gain = parent_H - perm_child_H
        null_gains.append(perm_gain)
        
    # --- C. P-Value ---
    null_gains = np.array(null_gains)
    # p = (r + 1) / (n + 1)
    p_val = (np.sum(null_gains >= real_gain) + 1) / (n_perms + 1)
    
    print(f"   [Info Gain] Real: {real_gain:.3f} bits | Null Max: {null_gains.max():.3f} | P-Val: {p_val:.3f}")
    
    return p_val

# --- 2. Splitting Logic (Elbow & GMM) ---

def get_elbow_columns(mi_matrix, exclusion_distance=5):
    L = mi_matrix.shape[0]
    mask = np.triu(np.ones((L, L), dtype=bool), k=exclusion_distance + 1)
    rows, cols = np.where(mask)
    scores = mi_matrix[rows, cols]
    
    # Sort scores descending
    sorted_indices = np.argsort(scores)[::-1]
    
    # Cap analysis window
    cap = min(100, len(sorted_indices))
    sorted_indices = sorted_indices[:cap]
    sorted_scores = scores[sorted_indices]
    
    if len(sorted_scores) < 3:
        # Fallback if almost no MI scores exist
        sel_rows = rows[sorted_indices]
        sel_cols = cols[sorted_indices]
        return np.unique(np.concatenate([sel_rows, sel_cols]))

    # Geometric Elbow
    n_points = len(sorted_scores)
    x_norm = np.linspace(0, 1, n_points)
    y_norm = (sorted_scores - sorted_scores.min()) / (sorted_scores.max() - sorted_scores.min() + 1e-9)
    
    line_vec = np.array([1, y_norm[-1] - y_norm[0]])
    line_vec = line_vec / np.linalg.norm(line_vec)
    vec_from_start = np.stack([x_norm, y_norm - y_norm[0]], axis=1)
    distances = np.abs(vec_from_start[:, 0] * line_vec[1] - vec_from_start[:, 1] * line_vec[0])
    
    elbow_idx = np.argmax(distances)
    n_selected = max(1, elbow_idx // 2) # Conservative selection
    selected_indices = sorted_indices[:n_selected]
    
    sel_rows = rows[selected_indices]
    sel_cols = cols[selected_indices]
    return np.unique(np.concatenate([sel_rows, sel_cols]))

def perform_pca_gmm_split(msa_int, vocab_size, col_indices, confidence_threshold=0.90):
    msa_subset = msa_int[:, col_indices]
    subset_one_hot = np.eye(vocab_size)[msa_subset]
    features = subset_one_hot.reshape(msa_int.shape[0], -1)
    
    n_comps = min(2, features.shape[1])
    pca = PCA(n_components=n_comps)
    pca_coords = pca.fit_transform(features)
    
    gmm = GaussianMixture(n_components=2, covariance_type='full', random_state=42)
    gmm.fit(pca_coords)
    
    probs = gmm.predict_proba(pca_coords) 
    max_probs = np.max(probs, axis=1)
    labels = gmm.predict(pca_coords)
    
    if len(np.unique(labels)) < 2:
        ambiguous_mask = max_probs < confidence_threshold
    else:
        sil_samples = silhouette_samples(pca_coords, labels)
        ambiguous_mask = (max_probs < confidence_threshold) | (sil_samples < 0.05)

    labels[ambiguous_mask] = -1 
    
    valid_mask = ~ambiguous_mask
    valid_labels = labels[valid_mask]
    valid_coords = pca_coords[valid_mask]
    
    if len(valid_labels) > 0 and len(np.unique(valid_labels)) >= 2:
        sil_score = silhouette_score(valid_coords, valid_labels)
    else:
        sil_score = -1 
    
    return labels, sil_score, pca_coords

def visualize_split(msa_int, labels, col_indices, alphabet="ACGTN-"):
    msa_subset = msa_int[:, col_indices]
    sort_idx = np.argsort(labels)
    sorted_msa = msa_subset[sort_idx]
    sorted_labels = labels[sort_idx]
    
    colors = ['#32CD32', '#4169E1', '#FFA500', '#DC143C', '#808080', '#FFFFFF']
    cmap_dna = ListedColormap(colors[:len(alphabet)])

    # fig, (ax_lbl, ax_msa) = plt.subplots(1, 2, figsize=(12, 6), 
    #                                      gridspec_kw={'width_ratios': [1, 20], 'wspace': 0.05})
    
    # label_cmap = ListedColormap(['#d3d3d3', '#1f77b4', '#ff7f0e']) 
    # ax_lbl.imshow(sorted_labels[:, None], aspect='auto', cmap=label_cmap, interpolation='nearest')
    # ax_lbl.set_title("ID", fontsize=10)
    # ax_lbl.axis('off')
    
    # im = ax_msa.imshow(sorted_msa, aspect='auto', interpolation='nearest', cmap=cmap_dna, vmin=0, vmax=len(alphabet)-1)
    
    # ax_msa.set_title(f"Informative Columns Only ({len(col_indices)} cols)", fontsize=12)
    # ax_msa.set_xlabel("Original Column Index")
    # ax_msa.set_xticks(np.arange(len(col_indices)))
    # ax_msa.set_xticklabels(col_indices, rotation=90, fontsize=8)
    # ax_msa.set_yticks([]) 
    
    # cbar = plt.colorbar(im, ax=ax_msa, ticks=np.arange(len(alphabet)))
    # cbar.ax.set_yticklabels(list(alphabet))
    
    # plt.tight_layout()
    # plt.show()

# --- 3. Recursive Pipeline (Simplified) ---

def recursive_msa_split(msa_strings, p_threshold=0.01, alphabet="ACGTN-", exclusion_distance=5):
    if len(msa_strings) < 3:
        return [msa_strings]

    print(f"Processing cluster with {len(msa_strings)} sequences...")
    msa_int, vocab_size = encode_msa(msa_strings, alphabet)
    
    # 1. Calculate Real MI directly (No sims)
    real_one_hot, P_i = get_marginals(msa_int, vocab_size)
    mi_matrix = compute_mi_scores(real_one_hot, P_i)
    
    # 2. Identify Best Columns
    top_cols = get_elbow_columns(mi_matrix, exclusion_distance=exclusion_distance)
    
    if len(top_cols) == 0:
        print("-> No informative columns found. Stopping.")
        return [msa_strings]
        
    print(f"   Identified {len(top_cols)} candidate columns.")
    
    # 3. Propose Split
    labels, score, _ = perform_pca_gmm_split(msa_int, vocab_size, top_cols, confidence_threshold=0.90)
    
    if score < 0.1: # Basic geometric check
        print(f"-> Split geometry too poor (Sil={score:.2f}). Stopping.")
        return [msa_strings]

    # 4. Validate Split with Information Gain Test
    # This replaces the old Monte Carlo test. We check if *this specific split* # reduces entropy more than a random shuffle of the same size.
    ig_p_value = check_split_entropy_significance(msa_int, vocab_size, labels, n_perms=100)
    
    if ig_p_value > p_threshold:
        print(f"-> Split rejected! IG p-value {ig_p_value:.3f} > {p_threshold}. treating as single cluster.")
        return [msa_strings]

    print(f"-> Split ACCEPTED (Sil={score:.2f}, IG p={ig_p_value:.3f}). Recursing...")
    visualize_split(msa_int, labels, top_cols, alphabet)
    
    group_A = [s for i, s in enumerate(msa_strings) if labels[i] == 0]
    group_B = [s for i, s in enumerate(msa_strings) if labels[i] == 1]
    
    n_dropped = len(msa_strings) - (len(group_A) + len(group_B))
    if n_dropped > 0:
        print(f"   (Dropped {n_dropped} ambiguous reads)")
    
    results = []
    results.extend(recursive_msa_split(group_A, p_threshold, alphabet, exclusion_distance))
    results.extend(recursive_msa_split(group_B, p_threshold, alphabet, exclusion_distance))
    
    return results

# --- Usage ---
for _ in range(100):
    print("")
    plas2 = mutate_sequence(plas1, 0.05)
    
    plas1_bc = "ACGAAAAGCGCTTT"
    plas2_bc = "ACGAATAGCGCTTT"
    
    n1 = 10
    n2 = 10
    
    plas1_seqs = [mutate_sequence(plas1, 0.05) for _ in range(n1)]
    plas2_seqs = [mutate_sequence(plas2, 0.05) for _ in range(n2)]
    
    plas1_bcs = [mutate_sequence(plas1_bc, 0.05) for _ in range(n1)]
    plas2_bcs = [mutate_sequence(plas2_bc, 0.05) for _ in range(n2)]
    
    plas_seqs = plas1_seqs + plas2_seqs
    bc_seqs = plas1_bcs + plas2_bcs
    
    from spoa import poa
    _, msa_plasmids = poa(plas_seqs)
    
    bc_gap_penalty = -20
    
    _, msa_bcs = poa(bc_seqs, g=bc_gap_penalty, algorithm=2)
    msa_bcs
    
    msa = [a + "-"*10 + b for a, b in zip(msa_bcs, msa_plasmids)]
    final_clusters = recursive_msa_split(msa, p_threshold=0.05, exclusion_distance=1)
```
