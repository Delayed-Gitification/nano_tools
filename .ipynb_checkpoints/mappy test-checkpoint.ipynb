```python
import mappy as mp
import sys

def get_best_hit(seq, aligner):
    best = None
    for hit in aligner.map(seq):
        if hit.is_primary and hit.mapq > 0:
            best = hit
            print("{}\t{}\t{}\t{}".format(hit.ctg, hit.r_st, hit.r_en, hit.cigar_str, hit.strand))
    return best

def extract_region(aligner, read_seq, flank_5p, flank_3p, region_name="Region"):
    """
    Generic function to extract sequence between two flanking sequences.
    Returns: (extracted_seq, orientation_str) or (None, error_msg)
    """
    print("H555555")
    h5 = get_best_hit(flank_5p, aligner)
    print("H333333")
    # h3 = get_best_hit(flank_3p, aligner)

    if not h5 or not h3:
        return None, "Flanks not found"

    if h5.strand != h3.strand:
        return None, "Flank orientation mismatch"

    # --- Forward Strand ---
    if h5.strand == 1:
        if h5.r_en >= h3.r_st:
            return None, "Negative distance (overlap/wrong order)"
        return read_seq[h5.r_en : h3.r_st], "+"

    # --- Reverse Strand ---
    # In RC mode, the physical 3' flank appears *before* the 5' flank on the read
    else:
        if h3.r_en >= h5.r_st:
            return None, "Negative distance (overlap/wrong order in RC)"
        
        # Extract RC segment and reverse complement back
        rc_segment = read_seq[h3.r_en : h5.r_st]
        return mp.revcomp(rc_segment), "-"


def process_fastq(fastq_path, bc_5p, bc_3p, ins_5p, ins_3p, min_length_fastq=3000):
    """
    Iterates over a FASTQ, extracting Barcode and Insert for each read.
    """
    
    print(f"Processing: {fastq_path}")
    print("Read_ID\tBarcode_Status\tBarcode_Seq\tInsert_Status\tInsert_Seq")
    print("-" * 80)

    # Alignment parameters optimized for small flank matching
    aligner_params = {
        'k': 5,              # k-mer size (lower k is more sensitive for short flanks)
        'w': 1,               # minimizer window size
        'min_chain_score': 0, 
        'min_dp_score': 0
    }

    # mp.fastx_read handles fastq/fasta and gzip automatically
    i = 0
    for name, seq, qual in mp.fastx_read(fastq_path):
        
        if len(seq) < min_length_fastq:
            continue

        print(seq)
        
        i += 1
        
        # 1. Build an aligner for THIS specific read
        # The read becomes the 'reference' so we can map flanks to it
        try:
            a = mp.Aligner(seq=seq, **aligner_params)
        except Exception as e:
            # Skip very short reads that might fail indexing
            print(f"{name}\tError_Indexing\tNA\tError_Indexing\tNA")
            continue

        # 2. Extract Barcode
        bc_seq, bc_ori = extract_region(a, seq, bc_5p, bc_3p, "Barcode")
        
        # 3. Extract Insert
        # ins_seq, ins_ori = extract_region(a, seq, ins_5p, ins_3p, "Insert")

        # 4. Format Output
        # Helper to make output cleaner
        bc_out = bc_seq if bc_seq else "NA"
        ins_out = ins_seq if ins_seq else "NA"
        
        # Status strings (e.g. "+" or "Flanks not found")
        bc_status = bc_ori if bc_seq else bc_ori
        ins_status = ins_ori if ins_seq else ins_ori

        print(f"{bc_status}\t{bc_out}\t{ins_status}\t{ins_out}")

        if i > 0:
            break

# --- usage example ---

__name__ = "__main__"

if __name__ == "__main__":
    
    # Example flanking sequences (Change these to your actual adapters)
    # Ensure these are long enough (15bp+) for mappy to find them reliably
    barcode_5p_flank = "CAAATAGGACGAGACGCGC" 
    barcode_3p_flank = "CGTAAACTGGATCCGCAGG"
    
    insert_5p_flank  = "CTTGGTGCCAGCTTATCA"
    insert_3p_flank  = "cctatgaagtgctctagtcaagtttaact".upper()
    
    # Replace with your actual FASTQ file path
    # You can pass this via sys.argv if you prefer
    fastq_file = "/Users/ogw/Downloads/ris_plasmids/no_sample_id/20251208_1553_MN41644_AYO707_6ec906fc/fastq_pass/combined.fastq.gz" 
    
    # Run
    # (Commented out to prevent execution error in this chat environment)
    process_fastq(fastq_file, barcode_5p_flank, barcode_3p_flank, insert_5p_flank, insert_3p_flank)
```
